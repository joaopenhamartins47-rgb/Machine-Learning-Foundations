```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline

# Underfitting: modelo linear
model_underfit = make_pipeline(
    PolynomialFeatures(degree=1, include_bias=False),
    LinearRegression()
)

# Ajuste intermediário: modelo quadrático
model_good_fit = make_pipeline(
    PolynomialFeatures(degree=2, include_bias=False),
    LinearRegression()
)

# Maior complexidade: polinômio de grau 4
model_overfit = make_pipeline(
    PolynomialFeatures(degree=4, include_bias=False),
    LinearRegression()
)

# X_train: tamanhos dos imóveis
# y_train: preços correspondentes

model_underfit.fit(X_train, y_train)
model_good_fit.fit(X_train, y_train)
model_overfit.fit(X_train, y_train)
```

Como analisar qual modelo é o melhor

```python
from sklearn.metrics import mean_squared_error

models = {
    "Grau 1": model_underfit,
    "Grau 2": model_good_fit,
    "Grau 4": model_overfit
}

for name, model in models.items():
    pred_train = model.predict(X_train)
    pred_test = model.predict(X_test)

    rmse_train = mean_squared_error(
        y_train, pred_train
    ) ** 0.5

    rmse_test = mean_squared_error(
        y_test, pred_test
    ) ** 0.5

    print(
        f"{name}: "
        f"treino={rmse_train:.2f}, "
        f"teste={rmse_test:.2f}"
    )
```
