```python
# Ideia: Calcular a previsao do modelo, depois calcular o custo e por fim calcular o gradient descent e aplica-lo para ajudar a reduzir os custos e otimizar o modelo
```

```python
def compute_cost(x, y, w, b):
    m = x.shape[0]
    total_cost = 0

    for i in range(m):
        f_wb = w * x[i] + b
        cost = (f_wb - y[i]) ** 2
        total_cost += cost

    total_cost = total_cost / (2 * m)

    return total_cost
```

```python
def compute_gradient(x, y, w, b):
    m = x.shape[0]

    dj_dw = 0
    dj_db = 0

    for i in range(m):
        f_wb = w * x[i] + b
        error = f_wb - y[i]

        dj_dw += error * x[i]
        dj_db += error

    dj_dw = dj_dw / m
    dj_db = dj_db / m

    return dj_dw, dj_db
```

```python
def gradient_descent(x, y, w_in, b_in,cost_function, gradient_function,alpha, num_iters):
    J_history = []
    w_history = []

    for i in range(num_iters):
        # Calcula o gradiente usando os parâmetros atuais
        dj_dw, dj_db = gradient_function(
            x, y, w_in, b_in
        )

        # Atualiza os parâmetros
        w_in = w_in - alpha * dj_dw
        b_in = b_in - alpha * dj_db

        # Registra o custo e os parâmetros
        J_history.append(
            cost_function(x, y, w_in, b_in)
        )
        w_history.append([w_in, b_in])

    return w_in, b_in, J_history, w_history
```
